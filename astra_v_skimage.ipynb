## Generate different shapes

```python
gen = PhantomGenerator(seed=1029, image_size=128)
P_all = gen.generate_triangle(n_samples=3_000)
rng = np.random.default_rng(0)
```

```python
n_id = rng.choice(3_000)
image = P_all[n_id].astype('float')
plt.matshow(image)
plt.savefig(os.path.join('figs', 'example_imgs', 'triangle_%d.png' % n_id), dpi=60)
```

Now we limit the angle to 30 degrees.

```python
gen = PhantomGenerator(seed=1029, image_size=128)
P_all = gen.generate_triangle(n_samples=3_000, max_angle=30)
rng = np.random.default_rng(0)
```

```python
n_id = rng.choice(3_000)
image = P_all[n_id].astype('float')
plt.matshow(image)
plt.savefig(os.path.join('figs', 'example_imgs', 'triangle-30d_%d.png' % n_id), dpi=60)
```
